```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/ADS_Project/data/raw/train.csv")

# Basic info
print(df.shape)
print(df.head())
print(df.info())
```

```python
# Convert date
df["date"] = pd.to_datetime(df["date"])

# Basic stats
print(df.describe())

# Plot one product from one store
sample = df[(df["store"] == 1) & (df["item"] == 1)]

plt.figure(figsize=(12,4))
plt.plot(sample["date"], sample["sales"])
plt.title("Sales Trend - Store 1 Item 1")
plt.show()
```
